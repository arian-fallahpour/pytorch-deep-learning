```python
# Random Tensors

# The way many NN learn is that they start with tensors with random numbers and adjust
#  those random numbers to better represent the data

#  random -> look at data -> update -> look at data -> update

random_tensor = torch.rand(3, 4) # Size (3, 4) with 2 dimensions

print(random_tensor)
print(random_tensor.shape)
print(random_tensor.ndim)
```

```python
random_image_size_tensor = torch.rand(size=(224, 224, 3)) # height, weight, color channel (RGB)
print(random_image_size_tensor.shape)
print(random_image_size_tensor.ndim)
```

```python
# Tensor with ones

ones = torch.ones(size=(3, 4))
print(ones)
print(ones.dtype)
```

```python
# Tensor data types

# Tensor datatypes is 1/3 big errors you will run into in PyTorch
# 1. Tensors not right datatype
# 2. Tensors not right shape
# 3. Tensors not on right device

float_32_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=None)
print(float_32_tensor.dtype)

int_32_tensor = torch.tensor([3, 6, 9], dtype=torch.long)
float_16_tensor = torch.tensor([3.0, 6.0, 9.0], dtype=torch.float16) # or torch.half
float_32_tensor = torch.tensor([3.0, 6.0, 9.0], 
                               dtype=None, # Data type (flaot32, float16, float64)
                               device=None, # None uses CPU, "cuda" uses GPU
                               requires_grad=False) # Track gradients or not with this tensor's ops

print(float_16_tensor.dtype)
print(float_16_tensor * float_32_tensor)
print(float_32_tensor * int_32_tensor)
```

```python
# Getting information from tensors

some_tensor = torch.rand(size=(3, 4), device="cuda", dtype=torch.float16)
print(some_tensor)

print(f"Data type of tensor: {some_tensor.dtype}")
print(f"Shape of tensor: {some_tensor.shape}")
print(f"Device tensor is on: {some_tensor.device}")
```

```python
#  Manipulating tensors

# Addition, Substraction, Multiplication (element-wise), Division, Matrix multiplication

tensor = torch.tensor([1, 2, 3])
print(tensor + 10)
print(tensor * 10)
print(tensor - 10)

print(torch.mul(tensor, 10))
print(torch.add(tensor, 10))
```

```python
# Matrix multiplication

# Two ways:
# 1. Element-wise
# 2. Matrix multiplication (dot product)

print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}") # Element wise

print(tensor, "x", tensor)
print(f"Equals: {torch.matmul(tensor, tensor)}") #  1*1 + 2*2 + 3*3 = 14

value = 0
for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
print(f"By hand: {value}")

# Common errors in deep learning
# 1. The inner dimensions must match
#    (3, 2) @ (3, 2) won't work
#    (2, 3) @ (3, 2) works
#    (3, 2) @ (2, 3) works

# torch.matmul(torch.rand(3, 2), torch.rand(3, 2)) # error
torch.matmul(torch.rand(3, 2), torch.rand(2, 3))
torch.matmul(torch.rand(2, 3), torch.rand(3, 2))

# 2, The resulting matrix has the shape of the outer dimensions
# (2, 3) @ (3, 2) -> (2, 2)
# (3, 2) @ (2, 3) -> (3, 3)

torch.matmul(torch.rand(2, 3), torch.rand(3, 2)).shape
```

```python
# Transpose

# Switches the axis

tensor_A = torch.tensor([[1, 2],
                         [3, 4],
                         [5, 6]])
tensor_B = torch.tensor([[7, 10],
                         [8, 11],
                         [9, 12]])
print(tensor_A)
print(tensor_B)
print(tensor_B.T)

print(torch.mm(tensor_A, tensor_B.T)) # Alias same as torch.matmul(tensor_A, tensor_B)
print(tensor_A.shape, tensor_B.T.shape)
```

```python
# Tensor aggregation

x = torch.arange(0, 100, 10)
print(x)
print(x.dtype)

print(torch.min(x), x.min())
print(torch.max(x), x.max())
print(torch.mean(x.type(torch.float32)), x.type(torch.float32).mean())
print(torch.sum(x), x.sum())

# Positional min and max
print(x.argmin(), x[0])
print(x.argmax(), x[9])
```

```python
# Reshaping, stacking, squeezing and unsqueezing tensors

x = torch.arange(1., 10.)
print(x, x.shape)
```

```python
# Reshape - reshapes an input tensor to a defined shape
x_reshaped = x.reshape(9, 1)
print(x_reshaped)
print(x_reshaped.shape)

print(torch.arange(1., 11.).reshape(2, 5))
print(torch.arange(1., 11.).reshape(5, 2))
```

```python
# View - return a view of an input tensor of a certain shape, but keep the same memory as the original tensor
z = x.view(1, 9)

print(z, z.shape)

# Changing z changes x (b/c using same memory)
z[:, 0] = 5
print(z, x)
```

```python
# Stacking - combine multiple tensors on top of each other (vstack), or side-by-side (hstack)

print(torch.stack([x, x, x, x], dim=0)) # like v-stack
print(torch.stack([x, x, x, x], dim=1)) # like h-stack? not sure

print(torch.vstack([x, x, x, x]))
print(torch.hstack([x, x, x, x]))
```

```python
# Squeeze - removed all `1` dimensions from a tensor

x_squeezed = x_reshaped.squeeze()
print(x_reshaped, x_reshaped.shape)
print(x_squeezed, x_squeezed.shape)
```

```python
# Unsqueeze - add a `1` dimension to a target tensor at a specific dimension

print(x_squeezed.unsqueeze(dim=0), x_squeezed.unsqueeze(dim=0).shape)
print(x_squeezed.unsqueeze(dim=1), x_squeezed.unsqueeze(dim=1).shape)
```

```python
# Permute - return a view of the input with dimensions permuted (swapped) in a certain way

x = torch.rand(size=(224, 224, 3)) # [height, width, color_channels]
permuted = x.permute(2, 0, 1) # [color_channels, height, width]

print(x.shape)
print(permuted.shape)
```

```python
# Indexing (selecting data from tensors)

x = torch.arange(1, 10).reshape(1, 3, 3)
print(x, x.shape)
print(x[0])
print(x[0][0], x[0, 0])
print(x[0][0][0], x[0, 0, 0])
print(x[0][2][2])

print(x[:, 0]) # Select all of 0th dimension
print(x[:, :, 1]) # Select all values of 0th and 1st dimension but only index 1 of 2nd dimension
print(x[:, 1, 1]) # Select all values of 0th dimension but only 1 index value of 1st and 2nd dimension
print(x[0, 0, :]) # Select index 0 of 0th and 1st dimension and all values in 2nd dimension
```

```python
# Pytorch tensors and Numpy

array = np.arange(1.0, 8.0)
tensor = torch.from_numpy(array)

print(array, tensor)
print(array.dtype, torch.arange(1.0, 8.0).dtype)

array = array + 1
print(array, tensor) # Changes do not propagate, it is a new tensor in memory

numpy_tensor = torch.ones(7).numpy()
print(numpy_tensor.dtype)
```

```python
# Reproducability

# No seed
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)

# Seed at start
RANDOM_SEED = 42
torch.manual_seed(RANDOM_SEED)

random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)

# Seed for every block
torch.manual_seed(RANDOM_SEED)
random_tensor_A = torch.rand(3, 4)
torch.manual_seed(RANDOM_SEED)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python
# Different ways of accessing a GPU in Pytorch

device = "cuda" if torch.cuda.is_available() else "cpu"

print(device)
print(torch.cuda.is_available())
print(torch.cuda.device_count())

# uses cpu
tensor = torch.tensor([1, 2, 3])
print(tensor, tensor.device)

# uses gpu
tensor_on_gpu = tensor.to(device)
print(tensor_on_gpu)

# move tensor back to cpu
tensor_back_on_cpu = tensor_on_gpu.cpu().numpy()
print(tensor_back_on_cpu)
```
